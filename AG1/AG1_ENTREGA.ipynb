## 1. Torres de Hanoi mediante divide y vencerás

El problema de las Torres de Hanoi consiste en mover `N` discos desde una torre de origen hasta una torre de destino usando una torre auxiliar.

Reglas:

- Solo se puede mover un disco cada vez.
- Nunca puede colocarse un disco grande encima de uno pequeño.

Este problema se resuelve por **divide y vencerás** porque el problema de mover `N` discos se reduce a mover `N-1` discos, mover el disco mayor y volver a mover `N-1` discos.

La recurrencia es:

$$T(1)=1,\qquad T(n)=2T(n-1)+1$$

Por tanto, el número de movimientos es:

$$T(n)=2^n-1$$

La complejidad temporal es:

$$O(2^n)$$

```python
def torres_hanoi(n, origen, destino, auxiliar, movimientos=None):
    """
    Resuelve el problema de las Torres de Hanoi mediante divide y vencerás.

    Parámetros:
    - n: número de discos.
    - origen: torre de origen.
    - destino: torre de destino.
    - auxiliar: torre auxiliar.
    - movimientos: lista donde se almacenan los movimientos.

    Devuelve:
    - Lista de movimientos realizados.
    """
    if movimientos is None:
        movimientos = []

    if n == 1:
        movimientos.append((origen, destino))
    else:
        torres_hanoi(n - 1, origen, auxiliar, destino, movimientos)
        movimientos.append((origen, destino))
        torres_hanoi(n - 1, auxiliar, destino, origen, movimientos)

    return movimientos


# Ejemplo de uso
N = 3
movimientos = torres_hanoi(N, "Torre 1", "Torre 3", "Torre 2")

print(f"Número de movimientos para {N} discos:", len(movimientos))
for i, mov in enumerate(movimientos, start=1):
    print(f"{i}. Mover disco de {mov[0]} a {mov[1]}")
```

## 2. Sucesión de Fibonacci

Aunque el ejemplo central de programación dinámica de la AG1 es el viaje por el río, también resulta útil ver Fibonacci para observar por qué conviene evitar cálculos repetidos.

La versión recursiva sencilla recalcula muchas veces los mismos valores. En cambio, con programación dinámica se almacenan los resultados intermedios.

```python
def fibonacci_recursivo(n):
    """
    Calcula el término n de Fibonacci de forma recursiva.
    Se usa la definición habitual:
    F(0)=0, F(1)=1.
    """
    if n < 0:
        raise ValueError("n debe ser no negativo")
    if n <= 1:
        return n
    return fibonacci_recursivo(n - 1) + fibonacci_recursivo(n - 2)


def fibonacci_dinamico(n):
    """
    Calcula el término n de Fibonacci mediante programación dinámica.
    Evita repetir cálculos guardando resultados anteriores.
    """
    if n < 0:
        raise ValueError("n debe ser no negativo")
    if n <= 1:
        return n

    tabla = [0] * (n + 1)
    tabla[0] = 0
    tabla[1] = 1

    for i in range(2, n + 1):
        tabla[i] = tabla[i - 1] + tabla[i - 2]

    return tabla[n]


n = 10
print("Fibonacci recursivo:", fibonacci_recursivo(n))
print("Fibonacci dinámico:", fibonacci_dinamico(n))
```

## 3. Devolución de cambio mediante técnica voraz

El problema consiste en devolver una cantidad usando monedas de distintos valores.

La estrategia **voraz** consiste en elegir en cada etapa la moneda de mayor valor posible sin superar la cantidad pendiente.

Funciona bien con muchos sistemas monetarios habituales, pero no siempre garantiza la solución óptima para cualquier conjunto de monedas.

```python
def cambio_monedas(cantidad, sistema):
    """
    Devuelve una solución al problema del cambio usando técnica voraz.

    Parámetros:
    - cantidad: cantidad que se desea formar.
    - sistema: lista de monedas ordenadas de mayor a menor.

    Devuelve:
    - Lista con el número de monedas de cada tipo.
    """
    solucion = [0] * len(sistema)
    valor_acumulado = 0

    for i, moneda in enumerate(sistema):
        monedas = (cantidad - valor_acumulado) // moneda
        solucion[i] = monedas
        valor_acumulado += monedas * moneda

        if valor_acumulado == cantidad:
            break

    return solucion


# Ejemplo con sistema monetario habitual
sistema = [25, 10, 5, 1]
cantidad = 27
solucion = cambio_monedas(cantidad, sistema)

print("Sistema monetario:", sistema)
print("Cantidad:", cantidad)
print("Solución voraz:", solucion)

for moneda, unidades in zip(sistema, solucion):
    if unidades > 0:
        print(f"{unidades} moneda(s) de {moneda}")


# Ejemplo donde el algoritmo voraz no da la solución óptima
sistema_no_canonico = [11, 5, 1]
cantidad = 15
solucion_voraz = cambio_monedas(cantidad, sistema_no_canonico)

print("\nSistema no canónico:", sistema_no_canonico)
print("Cantidad:", cantidad)
print("Solución voraz:", solucion_voraz)
print("Interpretación: 11 + 1 + 1 + 1 + 1 = 15, usa 5 monedas")
print("Pero la solución óptima sería: 5 + 5 + 5 = 15, usa 3 monedas")
```

## 4. Problema de las N-Reinas mediante vuelta atrás (*backtracking*)

El problema consiste en colocar `N` reinas en un tablero `N x N` sin que ninguna ataque a otra.

Representación usada:

$$
S = [s_1, s_2, \ldots, s_N]
$$

donde cada posición de la lista representa una columna y el valor almacenado representa la fila en la que se coloca la reina.

Una solución parcial se descarta si:

- dos reinas están en la misma fila;
- dos reinas están en la misma diagonal.

La técnica de vuelta atrás construye la solución por etapas y retrocede cuando una rama deja de ser prometedora.

```python
def es_prometedora(solucion, etapa):
    """
    Comprueba si la solución parcial es prometedora hasta la etapa indicada.

    solucion[columna] = fila
    """
    for i in range(etapa):
        # Misma fila
        if solucion[i] == solucion[etapa]:
            return False

        # Misma diagonal
        if abs(solucion[i] - solucion[etapa]) == abs(i - etapa):
            return False

    return True


def resolver_n_reinas(n, solucion=None, etapa=0, soluciones=None):
    """
    Resuelve el problema de las N-Reinas mediante backtracking.

    Devuelve todas las soluciones encontradas.
    """
    if solucion is None:
        solucion = [0] * n
    if soluciones is None:
        soluciones = []

    for fila in range(1, n + 1):
        solucion[etapa] = fila

        if es_prometedora(solucion, etapa):
            if etapa == n - 1:
                soluciones.append(solucion.copy())
            else:
                resolver_n_reinas(n, solucion, etapa + 1, soluciones)

    solucion[etapa] = 0
    return soluciones


def imprimir_tablero(solucion):
    """
    Imprime una solución del problema de las N-Reinas.
    """
    n = len(solucion)

    for fila in range(1, n + 1):
        for columna in range(n):
            if solucion[columna] == fila:
                print(" Q ", end="")
            else:
                print(" - ", end="")
        print()


# Ejemplo con 4 reinas
n = 4
soluciones = resolver_n_reinas(n)

print(f"Número de soluciones para {n} reinas:", len(soluciones))
print("Primera solución encontrada:", soluciones[0])
imprimir_tablero(soluciones[0])
```

## 5. Viaje por el río mediante programación dinámica

El problema consiste en calcular el coste mínimo para viajar entre embarcaderos.

Se parte de una matriz `TARIFAS`, donde `TARIFAS[i][j]` representa el precio de ir directamente desde el embarcadero `i` hasta el embarcadero `j`.

Cuando no existe conexión directa se usa un valor muy alto, por ejemplo `999` o `float("inf")`.

La programación dinámica guarda el mejor coste parcial entre cada par de embarcaderos. La idea es:

$$
P(i,j)=\min\{T(i,j), P(i,k)+T(k,j)\}
$$

para posibles embarcaderos intermedios `k`.

```python
INF = float("inf")

TARIFAS = [
    [0,   5,   4,   3, INF, INF, INF],
    [INF, 0, INF,   2,   3, INF,  11],
    [INF, INF, 0,   1, INF,   4,  10],
    [INF, INF, INF, 0,   5,   6,   9],
    [INF, INF, INF, INF, 0, INF,   4],
    [INF, INF, INF, INF, INF, 0,   3],
    [INF, INF, INF, INF, INF, INF, 0]
]


def precios_minimos(tarifas):
    """
    Calcula los precios mínimos entre embarcaderos mediante programación dinámica.

    Devuelve:
    - precios: matriz con los costes mínimos.
    - ruta: matriz con el embarcadero intermedio óptimo.
    """
    n = len(tarifas)

    precios = [fila.copy() for fila in tarifas]
    ruta = [["" for _ in range(n)] for _ in range(n)]

    for i in range(n):
        for j in range(n):
            if i != j and precios[i][j] < INF:
                ruta[i][j] = i

    # Como el grafo está orientado hacia embarcaderos posteriores, recorremos por distancia
    for distancia in range(1, n):
        for i in range(n - distancia):
            j = i + distancia

            for k in range(i + 1, j + 1):
                if precios[i][k] + precios[k][j] < precios[i][j]:
                    precios[i][j] = precios[i][k] + precios[k][j]
                    ruta[i][j] = k

    return precios, ruta


def reconstruir_ruta(ruta, desde, hasta):
    """
    Reconstruye la ruta óptima desde un embarcadero inicial hasta uno final.
    """
    if desde == hasta:
        return [desde]

    intermedio = ruta[desde][hasta]

    if intermedio == "" or intermedio == desde:
        return [desde, hasta]

    izquierda = reconstruir_ruta(ruta, desde, intermedio)
    derecha = reconstruir_ruta(ruta, intermedio, hasta)

    return izquierda[:-1] + derecha


precios, ruta = precios_minimos(TARIFAS)

desde = 0
hasta = 6

print("Precio mínimo desde", desde, "hasta", hasta, ":", precios[desde][hasta])
print("Ruta óptima:", reconstruir_ruta(ruta, desde, hasta))
```

# Práctica individual / reto: encontrar los dos puntos más cercanos

El reto consiste en encontrar los dos puntos más cercanos de un conjunto dado.

Se trabaja en tres niveles:

1. Caso 1D: puntos representados como números.  
2. Caso 2D: puntos representados como pares `(x, y)`.  
3. Caso 3D: puntos representados como ternas `(x, y, z)`.  

Primero se aplica fuerza bruta y después se usa una estrategia de divide y vencerás.

## 6. Reto en 1D: fuerza bruta

En 1D, los puntos son simplemente números.

La fuerza bruta compara todos los pares posibles. Si hay `n` puntos, el número de comparaciones es aproximadamente:

$$
\frac{n(n-1)}{2}
$$

Por tanto, la complejidad es:

$$
O(n^2)
$$

```python
import math
import random
from itertools import combinations

random.seed(42)


def distancia_1d(a, b):
    """Distancia entre dos puntos en una dimensión."""
    return abs(a - b)


def puntos_mas_cercanos_1d_fuerza_bruta(puntos):
    """
    Encuentra los dos puntos más cercanos en 1D mediante fuerza bruta.
    Complejidad: O(n^2).
    """
    if len(puntos) < 2:
        raise ValueError("Se necesitan al menos dos puntos")

    mejor_par = None
    mejor_distancia = float("inf")

    for p1, p2 in combinations(puntos, 2):
        d = distancia_1d(p1, p2)

        if d < mejor_distancia:
            mejor_distancia = d
            mejor_par = (p1, p2)

    return mejor_par, mejor_distancia


puntos_1d = random.sample(range(0, 10000), 12)
par, distancia = puntos_mas_cercanos_1d_fuerza_bruta(puntos_1d)

print("Puntos 1D:", puntos_1d)
print("Par más cercano:", par)
print("Distancia:", distancia)
```

## 7. Reto en 1D: divide y vencerás

En 1D se puede mejorar claramente.

La idea es:

1. Ordenar los puntos.
2. Una vez ordenados, los dos puntos más cercanos tienen que estar juntos en la lista ordenada.
3. Basta comparar puntos consecutivos.

La ordenación cuesta:

$$
O(n \log n)
$$

y recorrer la lista ordenada cuesta:

$$
O(n)
$$

Por tanto, la complejidad total queda dominada por la ordenación:

$$
O(n \log n)
$$

```python
def puntos_mas_cercanos_1d_divide_venceras(puntos):
    """
    Encuentra los dos puntos más cercanos en 1D usando ordenación.
    Aunque la comparación final es lineal, el coste dominante es ordenar.
    Complejidad: O(n log n).
    """
    if len(puntos) < 2:
        raise ValueError("Se necesitan al menos dos puntos")

    puntos_ordenados = sorted(puntos)

    mejor_par = (puntos_ordenados[0], puntos_ordenados[1])
    mejor_distancia = distancia_1d(puntos_ordenados[0], puntos_ordenados[1])

    for i in range(1, len(puntos_ordenados) - 1):
        d = distancia_1d(puntos_ordenados[i], puntos_ordenados[i + 1])

        if d < mejor_distancia:
            mejor_distancia = d
            mejor_par = (puntos_ordenados[i], puntos_ordenados[i + 1])

    return mejor_par, mejor_distancia


par_fb, distancia_fb = puntos_mas_cercanos_1d_fuerza_bruta(puntos_1d)
par_dv, distancia_dv = puntos_mas_cercanos_1d_divide_venceras(puntos_1d)

print("Resultado fuerza bruta:", par_fb, distancia_fb)
print("Resultado divide y vencerás:", par_dv, distancia_dv)
print("Coinciden las distancias:", distancia_fb == distancia_dv)
```

## 8. Reto en 2D: fuerza bruta

En 2D cada punto se representa como:

$$
(x,y)
$$

La distancia euclídea entre dos puntos es:

$$
d=\sqrt{(x_2-x_1)^2+(y_2-y_1)^2}
$$

La fuerza bruta vuelve a comparar todos los pares posibles, por lo que la complejidad es:

$$
O(n^2)
$$

```python
def distancia_euclidea(p1, p2):
    """
    Calcula la distancia euclídea entre dos puntos de cualquier dimensión.
    """
    return math.sqrt(sum((a - b) ** 2 for a, b in zip(p1, p2)))


def puntos_mas_cercanos_fuerza_bruta(puntos):
    """
    Encuentra los dos puntos más cercanos en cualquier dimensión mediante fuerza bruta.
    Complejidad: O(n^2).
    """
    if len(puntos) < 2:
        raise ValueError("Se necesitan al menos dos puntos")

    mejor_par = None
    mejor_distancia = float("inf")

    for p1, p2 in combinations(puntos, 2):
        d = distancia_euclidea(p1, p2)

        if d < mejor_distancia:
            mejor_distancia = d
            mejor_par = (p1, p2)

    return mejor_par, mejor_distancia


puntos_2d = [(random.randint(0, 10000), random.randint(0, 10000)) for _ in range(12)]

par_2d_fb, distancia_2d_fb = puntos_mas_cercanos_fuerza_bruta(puntos_2d)

print("Puntos 2D:", puntos_2d)
print("Par más cercano por fuerza bruta:", par_2d_fb)
print("Distancia:", distancia_2d_fb)
```

## 9. Reto en 2D: divide y vencerás

En 2D el método de divide y vencerás sigue esta idea:

1. Ordenar los puntos por la coordenada `x`.
2. Dividir el conjunto en dos mitades.
3. Resolver recursivamente en la mitad izquierda y en la mitad derecha.
4. Tomar la mejor distancia obtenida, llamada `delta`.
5. Revisar solo los puntos cercanos a la línea de división, porque un par que mejore `delta` no puede estar muy lejos de esa línea.

Esta implementación mantiene listas ordenadas por `x` y por `y`, y utiliza la idea de la franja central. Su complejidad esperada es:

$$
O(n \log n)
$$

```python
def closest_pair_2d_divide_and_conquer(puntos):
    """
    Encuentra los dos puntos más cercanos en 2D mediante divide y vencerás.

    La función es exacta y compara su resultado con fuerza bruta en tamaños pequeños.
    Complejidad: O(n log n), usando listas ordenadas por x y por y.
    """
    if len(puntos) < 2:
        raise ValueError("Se necesitan al menos dos puntos")

    # Si hay dos puntos exactamente iguales, la distancia mínima es 0.
    vistos = set()
    for p in puntos:
        if p in vistos:
            return (p, p), 0.0
        vistos.add(p)

    puntos_x = sorted(puntos, key=lambda p: (p[0], p[1]))
    puntos_y = sorted(puntos, key=lambda p: (p[1], p[0]))

    def rec(px, py):
        n = len(px)

        if n <= 3:
            return puntos_mas_cercanos_fuerza_bruta(px)

        mitad = n // 2
        x_medio = px[mitad][0]

        izquierda_x = px[:mitad]
        derecha_x = px[mitad:]

        conjunto_izquierda = set(izquierda_x)
        izquierda_y = [p for p in py if p in conjunto_izquierda]
        derecha_y = [p for p in py if p not in conjunto_izquierda]

        par_izq, dist_izq = rec(izquierda_x, izquierda_y)
        par_der, dist_der = rec(derecha_x, derecha_y)

        if dist_izq <= dist_der:
            mejor_par = par_izq
            delta = dist_izq
        else:
            mejor_par = par_der
            delta = dist_der

        franja = [p for p in py if abs(p[0] - x_medio) < delta]

        # En el algoritmo clásico de 2D basta revisar un número acotado de vecinos en y.
        for i in range(len(franja)):
            for j in range(i + 1, min(i + 8, len(franja))):
                d = distancia_euclidea(franja[i], franja[j])

                if d < delta:
                    delta = d
                    mejor_par = (franja[i], franja[j])

        return mejor_par, delta

    return rec(puntos_x, puntos_y)


par_2d_dv, distancia_2d_dv = closest_pair_2d_divide_and_conquer(puntos_2d)

print("Resultado fuerza bruta:", par_2d_fb, distancia_2d_fb)
print("Resultado divide y vencerás:", par_2d_dv, distancia_2d_dv)
print("Coinciden las distancias:", abs(distancia_2d_fb - distancia_2d_dv) < 1e-9)
```

## 10. Reto en 3D: extensión del problema

En 3D cada punto se representa como:

$$
(x,y,z)
$$

La distancia euclídea se calcula de forma análoga:

$$
d=\sqrt{(x_2-x_1)^2+(y_2-y_1)^2+(z_2-z_1)^2}
$$

Primero se resuelve por fuerza bruta y después se plantea una extensión de divide y vencerás usando la misma idea general:

1. Ordenar por la primera coordenada.
2. Dividir en dos mitades.
3. Resolver recursivamente en cada mitad.
4. Revisar solo la franja de puntos que puede mejorar la mejor distancia obtenida.

En dimensiones superiores la franja central puede requerir más comparaciones que en 2D, por lo que esta extensión es exacta y didáctica, pero no pretende ser la versión geométrica más optimizada posible.

```python
puntos_3d = [
    (random.randint(0, 10000), random.randint(0, 10000), random.randint(0, 10000))
    for _ in range(12)
]

par_3d_fb, distancia_3d_fb = puntos_mas_cercanos_fuerza_bruta(puntos_3d)

print("Puntos 3D:", puntos_3d)
print("Par más cercano por fuerza bruta:", par_3d_fb)
print("Distancia:", distancia_3d_fb)
```

```python
def puntos_mas_cercanos_nd_divide_venceras(puntos, dimension=0):
    """
    Extensión exacta de divide y vencerás para puntos de cualquier dimensión.

    La división se realiza por una coordenada y se comparan los candidatos
    que quedan dentro de la franja central.

    Observación:
    - En 1D y 2D existen versiones muy eficientes.
    - En 3D esta implementación mantiene la idea de divide y vencerás, pero la franja
      puede contener más comparaciones en el peor caso.
    """
    if len(puntos) < 2:
        raise ValueError("Se necesitan al menos dos puntos")

    dim_total = len(puntos[0])

    # Si hay dos puntos exactamente iguales, la distancia mínima es 0.
    vistos = set()
    for p in puntos:
        if p in vistos:
            return (p, p), 0.0
        vistos.add(p)

    def rec(lista, dim):
        n = len(lista)

        if n <= 3:
            return puntos_mas_cercanos_fuerza_bruta(lista)

        # En cada llamada se ordena por la coordenada que se está usando para dividir.
        lista = sorted(lista, key=lambda p: p[dim])

        mitad = n // 2
        valor_medio = lista[mitad][dim]

        izquierda = lista[:mitad]
        derecha = lista[mitad:]

        par_izq, dist_izq = rec(izquierda, (dim + 1) % dim_total)
        par_der, dist_der = rec(derecha, (dim + 1) % dim_total)

        if dist_izq <= dist_der:
            mejor_par = par_izq
            delta = dist_izq
        else:
            mejor_par = par_der
            delta = dist_der

        # Solo pueden mejorar delta los puntos próximos al plano de división
        franja = [p for p in lista if abs(p[dim] - valor_medio) < delta]

        # Comparación exacta dentro de la franja.
        for p1, p2 in combinations(franja, 2):
            d = distancia_euclidea(p1, p2)

            if d < delta:
                delta = d
                mejor_par = (p1, p2)

        return mejor_par, delta

    return rec(list(puntos), dimension)


par_3d_dv, distancia_3d_dv = puntos_mas_cercanos_nd_divide_venceras(puntos_3d)

print("Resultado fuerza bruta:", par_3d_fb, distancia_3d_fb)
print("Resultado divide y vencerás 3D:", par_3d_dv, distancia_3d_dv)
print("Coinciden las distancias:", abs(distancia_3d_fb - distancia_3d_dv) < 1e-9)
```

```python
# Comprobación automática con varios conjuntos aleatorios

def comprobar_reto():
    """
    Comprueba que los métodos rápidos/divide y vencerás devuelven la misma distancia
    que fuerza bruta en varios casos aleatorios.
    """
    random.seed(123)

    # 1D
    for _ in range(10):
        puntos = random.sample(range(0, 10000), 30)
        _, d_fb = puntos_mas_cercanos_1d_fuerza_bruta(puntos)
        _, d_dv = puntos_mas_cercanos_1d_divide_venceras(puntos)
        assert abs(d_fb - d_dv) < 1e-9

    # 2D
    for _ in range(10):
        puntos = [(random.randint(0, 10000), random.randint(0, 10000)) for _ in range(30)]
        _, d_fb = puntos_mas_cercanos_fuerza_bruta(puntos)
        _, d_dv = closest_pair_2d_divide_and_conquer(puntos)
        assert abs(d_fb - d_dv) < 1e-9

    # 3D
    for _ in range(10):
        puntos = [
            (random.randint(0, 10000), random.randint(0, 10000), random.randint(0, 10000))
            for _ in range(30)
        ]
        _, d_fb = puntos_mas_cercanos_fuerza_bruta(puntos)
        _, d_dv = puntos_mas_cercanos_nd_divide_venceras(puntos)
        assert abs(d_fb - d_dv) < 1e-9

    print("Todas las comprobaciones han sido correctas.")


comprobar_reto()
```

# Conclusión final

La actividad guiada permite practicar varias estrategias fundamentales de diseño de algoritmos:

- **Divide y vencerás**, aplicada a Torres de Hanoi y al reto de puntos cercanos.
- **Voracidad**, aplicada al cambio de monedas.
- **Vuelta atrás**, aplicada al problema de las N-Reinas.
- **Programación dinámica**, aplicada al viaje por el río.

El reto de los puntos más cercanos muestra claramente cómo una primera solución por fuerza bruta puede ser correcta pero poco eficiente, y cómo el análisis de complejidad permite buscar alternativas más rápidas.
